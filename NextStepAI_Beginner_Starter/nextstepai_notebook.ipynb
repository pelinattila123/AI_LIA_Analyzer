Detta notebook visar hur man kan skapa ett litet projekt i Python.
Vi använder:

Målet är att göra det enkelt att förstå.

```python
# Cell 1: Importera bibliotek
import json  # Importera json för att läsa JSON-filer
import random  # Importera random för slumpmässigt tips
import pandas as pd  # Importera pandas för tabeller och data
import matplotlib.pyplot as plt  # Importera matplotlib för enkel grafik

print("Biblioteken är importerade.")  # Skriv ut ett meddelande när allt fungerar
```

```python
# Cell 2: Läs data från JSON-filen
with open("companies.json", "r", encoding="utf-8") as fil:  # Öppna databasen
    data = json.load(fil)  # Läs innehållet och spara det i variabeln data
if data:
    print("Databasen innehåller företag.")
else:
    print("Databasen är tom.")

print("Här är områdena i databasen:")  # Skriv ut rubrik

for omrade in data:  # Loopa igenom varje område i databasen
    print("-", omrade)  # Skriv ut namnet på området

    
```

```python
# Cell 3: Skapa klasser och använd arv

def visa_omraden(data):
    print("Tillgängliga områden:")
    
    for omrade in data:
        print("-", omrade)


visa_omraden(data)


class KarriarHjalp:
    # Skapa en basklass

    def visa_meddelande(self):
        print("Välkommen till NextStepAI!")


class LIAHjalp(KarriarHjalp):  # Skapa en klass som ärver från KarriarHjalp
    def hitta_lia(self, omrade):  # Metod som visar företag i ett valt område
        print(f"\nDu valde området: {omrade}")  # Skriv ut vilket område användaren valde
        for foretag in data.get(omrade, []):  # Loopar igenom företagen i området
            print("-", foretag["company"])  # Skriv ut företagets namn
            print("  ", foretag["description"])  # Skriv ut beskrivningen


class IntervjuTraning(KarriarHjalp):  # Skapa en klass för intervjuträning
    def ge_tip(self):  # Metod som ger ett slumpmässigt tips
        tips = [  # Skapa lista med tips
            "Var tydlig när du svarar.",  # Tips 1
            "Ge konkreta exempel.",  # Tips 2
            "Visa intresse för företaget."  # Tips 3
        ]
        print("\nSlumpmässigt tips:")  # Skriv ut rubrik
        print(random.choice(tips))  # Välj ett slumpmässigt tips


hjalp = KarriarHjalp()  # Skapa ett objekt av basklassen
lia = LIAHjalp()  # Skapa ett objekt av subklassen
intervju = IntervjuTraning()  # Skapa ett objekt av intervjuträning

hjalp.visa_meddelande()  # Anropa metod från basklassen
lia.hitta_lia("AI")  # Anropa metod för AI-området
intervju.ge_tip()  # Anropa metod för slumpmässigt tips
```

```python
# Cell 4: Gör en enkel tabell med pandas
rad = []  # Skapa en tom lista
for omrade, foretag_lista in data.items():  # Loopar igenom databasen
    for foretag in foretag_lista:  # Loopar igenom varje företag
        rad.append({  # Lägg till ett nytt dict i listan
            "Område": omrade,  # Skriv in området
            "Företag": foretag["company"],  # Skriv in företagets namn
            "Kompetenser": ", ".join(foretag["skills"])  # Skriv in kompetenser
        })

# Skapa en DataFrame från listan
company_df = pd.DataFrame(rad)
print(company_df)  # Skriv ut tabellen
```

```python
# Cell 5: Gör en enkel graf med matplotlib
antal_per_omrade = company_df["Område"].value_counts()  # Räknar antal företag per område

plt.figure(figsize=(6, 4))  # Skapa en figur
plt.bar(antal_per_omrade.index, antal_per_omrade.values, color=["blue", "green", "orange"])  # Rita staplar
plt.title("Antal företag per område")  # Sätt titel
plt.xlabel("Område")  # Sätt x-axelns namn
plt.ylabel("Antal")  # Sätt y-axelns namn
plt.show()  # Visa grafen
```

```python
# Hämta jobbannonser från Arbetsförmedlingens API
import requests

url = "https://jobsearch.api.jobtechdev.se/search?q=python&limit=5"

try:
    response = requests.get(url, timeout=10)
    response.raise_for_status()

    api_data = response.json()

    for jobb in api_data["hits"]:
        print("-", jobb["headline"])

except requests.exceptions.RequestException:
    print("Fel: kunde inte hämta jobbannonser från API:t.")

except ValueError:
    print("Fel: API:t skickade inte giltig JSON.")

except KeyError:
    print("Fel: API-svaret har inte rätt format.")
```
