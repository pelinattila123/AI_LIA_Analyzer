```python
import random
import json

# NEXTSTEPAI
# Skapat av Pelin Korkmaz
# ==========================================
# KLASSER
# ==========================================

class KarriarHjalp:

    def visa_meddelande(self):
        print("Välkommen till NextStepAI")

class CVAnalys(KarriarHjalp):

    def analysera(self):
        self.visa_meddelande()
        print("CV-analys startar")

class IntervjuTraning(KarriarHjalp):

    def starta(self):
        self.visa_meddelande()
        print("Intervjuträning startar")

# ==========================================
# FUNKTION: HITTA LIA
# ==========================================

def hitta_lia():

    print("Du valde Hitta LIA.")

    foretag_lista = ["Knowit", "Nexer", "AFRY", "Sopra Steria"]

    print("Alla företag i databasen:")
    for foretag in foretag_lista:
        print(foretag)

    omraden = {
        "1": {
            "foretag": ["Knowit", "Nexer"],
            "text": "Passar dig som är intresserad av AI."
        },
        "2": {
            "foretag": ["AFRY", "Knowit"],
            "text": "Passar dig som gillar Python-utveckling."
        },
        "3": {
            "foretag": ["Sopra Steria", "AFRY"],
            "text": "Passar dig som vill arbeta med dataanalys."
        }
    }

    print("\nVälj ett område:")
    print("1. AI")
    print("2. Python")
    print("3. Dataanalys")

    omrade = input("Välj område: ")

    if omrade in omraden:

        print("\nFöretag:")

        for foretag in omraden[omrade]["foretag"]:
            print(foretag)

        print(omraden[omrade]["text"])

    ansokan = input(
        "\nVill du se vilka kompetenser som ofta efterfrågas? (ja/nej): "
    )

    if ansokan.lower() == "ja":

        print("\nVanliga kompetenser:")
        print("- Python")
        print("- Git")
        print("- SQL")

    kompetens = input("\nVilken av dessa kompetenser kan du redan? ")

    print(f"Bra! Du har angett: {kompetens}")

    matchningar = {
        "python": ("80%", "Knowit"),
        "git": ("60%", "Nexer"),
        "sql": ("90%", "AFRY")
    }

    if kompetens.lower() in matchningar:

        matchning, foretag = matchningar[kompetens.lower()]

        print(f"Matchning: {matchning}")
        print(f"Rekommenderat företag: {foretag}")

    else:

        print("Matchning: 30%")
        print("Du behöver utveckla fler kompetenser.")

# ==========================================
# FUNKTION: CV-ANALYS
# ==========================================

def cv_analys():

    print("Du valde CV-analys.")

    try:

        cv_text = input(
            "Vilka kompetenser har du? (t.ex. Python, Git, SQL): "
        )

        print("\nCV-analys:")

        kompetenser = ["python", "git", "sql"]

        for kompetens in kompetenser:

            if kompetens in cv_text.lower():
                print(f"✓ {kompetens.capitalize()} hittades")

        print("\nAnalysresultat:")

        if "python" in cv_text.lower() and "git" in cv_text.lower():

            print("Matchning: 90%")
            print("Rekommenderat företag: Knowit")

        elif "python" in cv_text.lower():

            print("Matchning: 80%")
            print("Rekommenderat företag: Nexer")

        elif "sql" in cv_text.lower():

            print("Matchning: 70%")
            print("Rekommenderat företag: AFRY")

        else:

            print("Matchning: 30%")
            print("Du behöver utveckla fler kompetenser.")

    except:

        print("Ett fel uppstod.")

# ==========================================
# FUNKTION: INTERVJUTRÄNING
# ==========================================

def intervju_traning():

    print("Du valde Träna intervju.")

    tips_lista = [
        "Var tydlig när du svarar.",
        "Ge konkreta exempel.",
        "Visa intresse för företaget."
    ]

    print("Tips:", random.choice(tips_lista))

    svar1 = input("\nFråga 1: Varför vill du arbeta inom AI?\nSvar: ")
    svar2 = input("\nFråga 2: Vilka programmeringsspråk kan du?\nSvar: ")
    svar3 = input("\nFråga 3: Varför skulle vi välja dig som praktikant?\nSvar: ")

    print("\nTack för dina svar!")

    print("\n==============================")
    print("SAMMANFATTNING AV INTERVJUN")
    print("==============================")

    print(f"\nFråga 1: Varför vill du arbeta inom AI?\nSvar: {svar1}")
    print(f"\nFråga 2: Vilka programmeringsspråk kan du?\nSvar: {svar2}")
    print(f"\nFråga 3: Varför skulle vi välja dig som praktikant?\nSvar: {svar3}")

    print("\nBra jobbat!")
    print("Fortsätt öva inför din LIA-intervju.")

    data = {
    "fraga1": svar1,
    "fraga2": svar2,
    "fraga3": svar3
}

    with open("intervju.json", "w") as fil:
         json.dump(data, fil, indent=4)

    print("Intervjun har sparats i intervju.json")

# ==========================================
# HUVUDPROGRAM
# ==========================================

namn = input("Vad heter du? ")

print(f"\nVälkommen {namn}!")
print("NextStepAI är redo att hjälpa dig.")

print("\nVad vill du göra?")
print("1. Hitta LIA")
print("2. CV-analys")
print("3. Träna intervju")

val = input("Välj 1, 2 eller 3: ")

if val == "1":

    hitta_lia()

elif val == "2":

    cv = CVAnalys()
    cv.analysera()
    cv_analys()

elif val == "3":

    intervju_traning()

else:

    print("Ogiltigt val.")
```
